# HE-MedInfer — Results Reproduction Notebook
Reproduces Figures 4, 5, and 6 from the paper using simulated data.

```python
import numpy as np
import matplotlib.pyplot as plt
import sys, os
sys.path.insert(0, os.path.join(os.path.dirname(''), '..'))
from src.evaluation.evaluate import simulate_results, report

results = simulate_results()
report(results)
```

```python
# Figure 4 — Accuracy comparison
labels = ['Plaintext MLP', 'DP-SGD', 'Lee et al.', 'HE-MedInfer']
means  = [results['plain_acc'].mean(), results['dp_acc'].mean(),
           results['lee_acc'].mean(),  results['he_acc'].mean()]
errs   = [results['plain_acc'].std()*1.96, results['dp_acc'].std()*1.96,
           results['lee_acc'].std()*1.96,  results['he_acc'].std()*1.96]

fig, ax = plt.subplots(figsize=(6,4))
ax.bar(labels, [m*100 for m in means], yerr=[e*100 for e in errs],
       capsize=5, color=['#cccccc','#aaaaaa','#888888','#4472C4'])
ax.set_ylabel('Diagnostic Accuracy (%)')
ax.set_ylim(82, 97)
ax.set_title('Figure 4 — Accuracy under Encrypted and Plaintext Inference')
plt.tight_layout()
plt.savefig('../data/results/fig4_accuracy.pdf')
plt.show()
```
